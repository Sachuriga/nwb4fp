```python
import numpy as np
from scipy.stats import f

# --- 1. 基础辅助函数 (之前缺失的部分) ---
def get_circular_stats(data):
    """
    计算环形数据的平均角度 (Mean Angle) 和 向量长度 (R)。
    """
    if len(data) == 0:
        return np.nan, 0
        
    # 计算正弦和余弦的和
    sin_sum = np.sum(np.sin(data))
    cos_sum = np.sum(np.cos(data))
    n = len(data)
    
    # 平均角度 (arctan2 自动处理象限)
    mean_angle = np.arctan2(sin_sum, cos_sum)
    
    # 向量长度 R (0=随机, 1=完全一致)
    R = np.sqrt(sin_sum**2 + cos_sum**2) / n
    
    return mean_angle, R

# --- 2. 环形标准差 (文中提到的 SD) ---
def get_circular_sd(data):
    """
    计算环形标准差 (Circular Standard Deviation)。
    公式: sqrt(-2 * ln(R))
    """
    if len(data) == 0: return np.nan
    _, R = get_circular_stats(data) 
    
    if R == 0: return np.inf 
    # 注意：如果 R 非常小，这个值会很大
    return np.sqrt(-2 * np.log(R))

# --- 3. Watson-Williams Test (Circular ANOVA) ---
def estimate_kappa(R):
    """
    根据 R 值估算集中度参数 Kappa (用于 ANOVA 计算)。
    """
    if R < 0.53:
        return 2 * R + R**3 + (5 * R**5) / 6
    elif R < 0.85:
        return -0.4 + 1.39 * R + 0.43 / (1 - R)
    else:
        return 1 / (3 * R - 4 * R**2 + R**3)

def watson_williams_test(data1, data2):
    """
    执行 Watson-Williams F-test (即 Circular ANOVA)。
    比较两组数据的平均角度是否显著不同。
    """
    n1 = len(data1)
    n2 = len(data2)
    N = n1 + n2
    
    # 获取各自和合并后的 R 值
    _, R1 = get_circular_stats(data1)
    _, R2 = get_circular_stats(data2)
    _, R_combined = get_circular_stats(np.concatenate([data1, data2]))
    
    # 计算加权 R (Weighted Resultants)
    rw1 = R1 * n1
    rw2 = R2 * n2
    rw_comb = R_combined * N
    
    # 估算 Kappa (使用合并数据)
    kappa = estimate_kappa(R_combined)
    
    # 修正因子 (Correction factor, Mardia & Jupp, 2000)
    if kappa > 1:
        correction = 1 + 3 / (8 * kappa) 
    else:
        # 如果数据太散 (kappa < 1)，ANOVA 可能不准
        correction = 1 
        
    # 计算 F 统计量
    numerator = (n1 + n2 - 2) * ((rw1 + rw2) - rw_comb)
    denominator = (N - rw_comb)
    
    if denominator == 0: return np.nan, 1, N-2, 1.0 # 避免除以0
    
    F_stat = correction * (numerator / denominator)
    
    # 自由度
    df1 = 1       # 组数 - 1
    df2 = N - 2   # 总样本数 - 组数
    
    # 计算 P 值
    p_value = f.sf(F_stat, df1, df2)
    
    return F_stat, df1, df2, p_value

# --- 4. 运行示例 (使用你的变量名) ---

# 假设这里是你的真实数据变量 (请确保你已经加载了这些数据)
# 这里我用随机数据代替演示
data_ko = np.random.vonmises(0, 4, 50)       # 均值0，比较集中
data_het = np.random.vonmises(1, 4, 53)      # 均值1，比较集中 (有偏移)

# 1. 计算统计量 (Mean ± SD)
m_ko, _ = get_circular_stats(data_ko)
sd_ko = get_circular_sd(data_ko)

m_het, _ = get_circular_stats(data_het)
sd_het = get_circular_sd(data_het)

print(f"KO Group:  Mean={np.degrees(m_ko):.1f}°, SD={sd_ko:.2f} rad")
print(f"Het Group: Mean={np.degrees(m_het):.1f}°, SD={sd_het:.2f} rad")
print("-" * 30)

# 2. 运行 Circular ANOVA
F_val, df1, df2, p_val = watson_williams_test(data_ko, data_het)

print(f"Circular ANOVA Results:")
print(f"F({df1}, {df2}) = {F_val:.2f}")
print(f"p-value = {p_val:.5f}")

if p_val < 0.05:
    print("结论: 两组相位的平均角度存在显著差异 (Significant Difference)")
else:
    print("结论: 两组相位的平均角度没有显著差异")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from scipy.ndimage import gaussian_filter1d
from scipy.stats import chi2, f  # Added 'f' for F-test

# --- 1. HELPER FUNCTIONS (Pure Numpy/Scipy) ---

def get_circular_stats(data):
    """Calculates Mean Angle and Vector Length (R)."""
    if len(data) == 0: return np.nan, 0
    sin_sum = np.sum(np.sin(data))
    cos_sum = np.sum(np.cos(data))
    n = len(data)
    mean_angle = np.arctan2(sin_sum, cos_sum)
    R = np.sqrt(sin_sum**2 + cos_sum**2) / n
    return mean_angle, R

def get_circular_sd(data):
    """Calculates Circular Standard Deviation (sqrt(-2ln(R)))."""
    if len(data) == 0: return np.nan
    _, R = get_circular_stats(data)
    if R == 0: return np.inf
    # Result is in radians
    return np.sqrt(-2 * np.log(R))

def rayleigh_test_numpy(data):
    """Calculates Rayleigh Test P-value for uniformity."""
    if len(data) == 0: return 1.0
    n = len(data)
    _, R = get_circular_stats(data)
    z = n * R**2
    p_value = np.exp(-z) * (1 + (2*z - z**2) / (4*n) - (24*z - 132*z**2 + 76*z**3 - 9*z**4) / (288*n**2))
    return p_value

# --- Watson-Williams Test Helpers ---
def estimate_kappa(R):
    """Estimates concentration parameter kappa for ANOVA."""
    if R < 0.53: return 2 * R + R**3 + (5 * R**5) / 6
    elif R < 0.85: return -0.4 + 1.39 * R + 0.43 / (1 - R)
    else: return 1 / (3 * R - 4 * R**2 + R**3)

def watson_williams_test(data1, data2):
    """
    Performs Watson-Williams F-test (Circular ANOVA) for two samples.
    Returns: F-statistic, df1, df2, p-value
    """
    n1, n2 = len(data1), len(data2)
    N = n1 + n2
    if n1 == 0 or n2 == 0: return np.nan, 0, 0, np.nan

    _, R1 = get_circular_stats(data1)
    _, R2 = get_circular_stats(data2)
    _, R_comb = get_circular_stats(np.concatenate([data1, data2]))

    rw1, rw2, rw_comb = R1*n1, R2*n2, R_comb*N
    kappa = estimate_kappa(R_comb)
    
    # Correction factor (Mardia & Jupp)
    correction = 1 + 3 / (8 * kappa) if kappa > 1 else 1
    
    numerator = (N - 2) * ((rw1 + rw2) - rw_comb)
    denominator = (N - rw_comb)
    
    if denominator == 0: return 0, 1, N-2, 1.0

    F_stat = correction * (numerator / denominator)
    df1, df2 = 1, N - 2
    p_value = f.sf(F_stat, df1, df2)
    
    return F_stat, df1, df2, p_value

def add_vector_arrow(ax, data, color):
    """Adds a mean resultant vector arrow."""
    if len(data) == 0: return
    mean_angle, R = get_circular_stats(data)
    # Plot Arrow
    ax.annotate('', xy=(mean_angle, R), xytext=(0, 0),
                arrowprops=dict(facecolor=color, edgecolor='black', width=2, headwidth=8, alpha=0.9))
    # Plot Dashed Line
    ax.plot([mean_angle, mean_angle], [0, .5], color=color, linestyle='--', linewidth=1, alpha=0.5)

def plot_polar_area(ax, data, color, bins=36, alpha=0.5, sigma=1):
    hist, bin_edges = np.histogram(data, bins=bins, range=(0, 2*np.pi), density=True)
    hist_smoothed = gaussian_filter1d(hist, sigma=sigma, mode='wrap')
    theta = (bin_edges[:-1] + bin_edges[1:]) / 2
    ax.fill_between(theta, 0, hist_smoothed, color=color, alpha=alpha, edgecolor=color)

def compare_groups(name, data_ko, data_het):
    """Run stats and print to console in paper format."""
    print(f"\n=== {name} Statistics ===")
    
    # 1. Mean ± SD
    m_ko, _ = get_circular_stats(data_ko)
    sd_ko = get_circular_sd(data_ko)
    m_het, _ = get_circular_stats(data_het)
    sd_het = get_circular_sd(data_het)
    
    print(f"KO : Mean {np.degrees(m_ko):5.1f}° ± {sd_ko:.2f} rad (N={len(data_ko)})")
    print(f"Het: Mean {np.degrees(m_het):5.1f}° ± {sd_het:.2f} rad (N={len(data_het)})")
    
    # 2. Circular ANOVA
    F_val, df1, df2, p_val = watson_williams_test(data_ko, data_het)
    sig = "***" if p_val < 0.001 else "**" if p_val < 0.01 else "*" if p_val < 0.05 else "ns"
    print(f"Circular ANOVA: F({df1},{df2}) = {F_val:.2f}, p = {p_val:.4f} ({sig})")

# --- 2. EXECUTE PLOTS AND STATS ---

# Create figure
fig = plt.figure(figsize=(7.2, 2.4), dpi=1200) # Adjusted DPI for screen speed
plt.rcParams.update({'font.size': 7,'font.family': 'sans-serif','font.sans-serif':['Arial', 'Calibri','DejaVu Sans', 'sans-serif']})
plt.rcParams.update({
    'axes.labelpad': -8,
    'ytick.major.pad': -3,
    'xtick.major.pad': -3,
    'ytick.major.size': 2,
    'xtick.major.size': 2
})
gs = gridspec.GridSpec(1, 3, height_ratios=[1], width_ratios=[1,1,1])

ax1 = fig.add_subplot(gs[0, 0], projection='polar')
ax2 = fig.add_subplot(gs[0, 1], projection='polar')
ax3 = fig.add_subplot(gs[0, 2], projection='polar')

# NOTE: Assuming bin_num is defined in your variable list, e.g.:
bin_num = 36 

# --- DEEP LAYER ---
ax1.set_title('Deep')
ax1.set_theta_zero_location('N')
ax1.set_theta_direction(-1)
plot_polar_area(ax1, preferred_phase_ko_deep, 'blue', bins=bin_num)
plot_polar_area(ax1, preferred_phase_het_deep, 'red', bins=bin_num)
add_vector_arrow(ax1, preferred_phase_ko_deep, 'blue')
add_vector_arrow(ax1, preferred_phase_het_deep, 'red')
# Run Stats
compare_groups("Deep Layer", preferred_phase_ko_deep, preferred_phase_het_deep)

# --- SUPERFICIAL LAYER ---
ax2.set_title('Superficial')
ax2.set_theta_zero_location('N')
ax2.set_theta_direction(-1)
plot_polar_area(ax2, preferred_phase_ko_superficial, 'blue', bins=bin_num)
plot_polar_area(ax2, preferred_phase_het_superficial, 'red', bins=bin_num)
add_vector_arrow(ax2, preferred_phase_ko_superficial, 'blue')
add_vector_arrow(ax2, preferred_phase_het_superficial, 'red')
# Run Stats
compare_groups("Superficial Layer", preferred_phase_ko_superficial, preferred_phase_het_superficial)

# --- INTERNEURONS ---
ax3.set_title('Interneurons')
ax3.set_theta_zero_location('N')
ax3.set_theta_direction(-1)
plot_polar_area(ax3, preferred_phase_ko_int, 'cyan', bins=bin_num)
plot_polar_area(ax3, preferred_phase_het_int, 'magenta', bins=bin_num)
add_vector_arrow(ax3, preferred_phase_ko_int, 'cyan')
add_vector_arrow(ax3, preferred_phase_het_int, 'magenta')
# Run Stats
compare_groups("Interneurons", preferred_phase_ko_int, preferred_phase_het_int)

plt.tight_layout()
plt.savefig(r'/Users/sachuriga/Desktop/Projects/CR_CA1_paper/Figures_mac/suppfig9.png', transparent=True, dpi=1200, bbox_inches='tight')
plt.show()
```
